## 1 · Lists

The most fundamental data container in Python is a **list** — an ordered,
mutable sequence of values. Lists are written with square brackets `[ ]`, and
elements are separated by commas.

### Creating a list

You can put any mix of values inside a list:

````python
# A list of four integers — note the square brackets and commas
l = [2, 3, 5, 7]
l   # typing the variable name on its own in a notebook prints its value
````

Lists can contain multiple data types as entries:

````python
# Lists can mix any data types in a single container:
m = [10,      # int
     'ES',    # str (string)
     -1.234,  # float
     True]    # bool
m
````

Lists can even hold other lists as entries (a **nested list** or matrix):

````python
# The third element is itself a list — this creates a nested list
J = [21, 5, ['apples', 12, False]]
J
````

**Creating an empty list** — useful when you want to build a list one element
at a time inside a loop:

````python
empty_list = []   # square brackets with nothing inside
empty_list        # calling a variable on its own prints its value in a notebook
````

You can also create an empty list with the built-in `list()` function:

````python
another_empty = list()   # list() with no arguments creates an empty list
another_empty            # prints [] to confirm it is empty
````

### Nested lists and matrix-style access

A nested list behaves like a 2-D matrix. To access an element you index twice:
first for the row, then for the column.

````python
# Each inner list is one row of the matrix:
B = [[1, 2],   # row 0:  first element is 1, second is 2
     [0, 4]]   # row 1:  first element is 0, second is 4
B
````

The element **2** is in the first row, second column.
Python uses 0-based indexing, so:

- row 0 → `B[0]` → `[1, 2]`
- element at row 0, column 1 → `B[0][1]`

````python
B[0][1]   # first row, second column — should be 2
````

### 0-based indexing

Python is a **0-indexed** language. This means if you want the *first* element of a list, you ask for index `0`, not `1`.

For a list with `n` elements, valid indices are `0, 1, 2, …, n-1`.

Use a **negative index** to count from the end:
- `-1` is the last element
- `-2` is the second-to-last, and so on.

````python
fruits = ['apple', 'banana', 'cherry', 'date']
print(fruits[0])    # first element
print(fruits[2])    # third element
print(fruits[-1])   # last element
print(fruits[-2])   # second-to-last
````

### Slicing list elements

A **slice** extracts a contiguous sub-sequence:

```
list[start : stop : step]
```

- `start` is included, `stop` is **excluded**
- `step` defaults to 1 (every element)

````python
sl = ['apples', 'pears', 'oranges', 'cherries', 'grapefruits', 'pomegranates']
sl
````

Retrieve elements at indices 1, 2, and 3 (stop at 4, which is excluded):

````python
sl[1:4]   # start at index 1 ('pears'), stop before index 4 ('cherries') → gives indices 1, 2, 3
````

Leaving `start` or `stop` blank tells Python to go all the way to the
beginning or end of the list:

````python
sl[2:]   # from index 2 to the very end
````

The **step** value controls how many positions to skip between picks.
`sl[1::2]` means: start at index 1, go to the end, take every second element:

````python
sl[1::2]   # start at index 1, no stop, step=2 → picks every other element: indices 1, 3, 5
````

### Finding the position of an element

`.index(value)` returns the position of the **first** occurrence of a value:

````python
sl.index('cherries')   # returns 3
````

If the same value appears multiple times, `.index()` only returns the first
hit. To find **all** positions you need a loop (try this as an exercise):

````python
repeat = [14, 15, 16, 16, 17]
repeat.index(16)   # returns 2 (the first 16)
````

### Replacing and adding list elements

Because lists are **mutable**, you can change any element after creation.
Use the assignment syntax `list[index] = new_value`:

````python
l = [2, 3, 5, 7]
l[1] = 'lantern'   # replace the element at index 1
l
````

To **add** an element to the end of a list, use `.append()`.
This modifies the list in place — no need to re-assign the variable:

````python
l = [1, 2, 3, 4]
l.append(5)   # add 5 to the END of the list — modifies l in place, returns None
l             # print the updated list
````

You can also **concatenate** two lists with `+`.
This creates a **new** list; the originals are unchanged:

````python
m = [1, 2, 3, 4]
newlist = m + [5]
newlist   # m itself is unchanged
````

Trying to add a plain number (not wrapped in a list) to a list raises a

````python
m + 5    # TypeError: can only concatenate list (not "int") to list
# Python cannot join a list and a bare integer.
# Fix: wrap the value in a list →  m + [5]
````

Multiply a list by an integer to repeat it that many times.
This is handy for initialising a list of fixed length:

````python
m * 3   # repeat the list m three times end-to-end; m itself is NOT changed
````

````python
ones = [1] * 50    # a list of 50 ones
print(ones)
````

### More useful list methods

| Method | What it does |

### Sorting lists

`sorted(list)` returns a **new** sorted list (does not modify the original).
`list.sort()` sorts **in place** (modifies the original, returns `None`).

````python
ln = [2, -5, 10, 7, 3, 5]
sorted(ln)   # returns a new sorted list; ln is unchanged
````

````python
lm = ['apples', 'pears', 'oranges', 'cherries', 'grapefruits', 'pomegranates']
sorted(lm)   # alphabetical
````

````python
lm.sort()    # in-place — lm itself is now sorted
lm
````

**Sorting mixed types:**
Python can sort a list that mixes `int`, `float`, and `bool` — because `bool`
is a subclass of `int` (`True == 1`, `False == 0`):

````python
test1 = [1, 3, 0, True, 10, 3.5, False]
sorted(test1)   # works — booleans treated as 0/1
````

But sorting a list that mixes `int` and `str` raises a `TypeError` — Python
cannot decide whether `3` is less than `'4'`:

````python
test2 = [1, '2', 3, '4', 5, 'A']
# sorted(test2)   # ← uncomment to see TypeError
````

To sort `test2`, we first separate the integers and strings, sort each group,
then recombine:

````python
# isinstance(x, int) returns True when x is an integer (not a string)
test2_ints    = sorted([x for x in test2 if isinstance(x, int)])    # collect + sort the ints
test2_strings = sorted([x for x in test2 if isinstance(x, str)])    # collect + sort the strings
test2_sorted  = test2_ints + test2_strings   # concatenate: ints first, then strings
print(test2_sorted)
````

### Mutability and the aliasing trap

Lists are **mutable** — you can change their contents after creation.
This is powerful, but introduces a subtle trap: **assignment does not copy a
list**. It creates a second name that points to the same underlying object.

````python
my_list = [1, 2, 3]
a = my_list          # a is NOT a copy — it is another name for the same list

my_list[1] = 5       # modify through one name …

print(my_list)       # [1, 5, 3]
print(a)             # [1, 5, 3]  ← a changed too!
````

> **Why did `a` change?**
> `a = my_list` makes `a` point to the **same list in memory**.
> Changing one changes the other, because they refer to the same object.

This is called **aliasing** and is one of the most common sources of subtle
bugs in Python. The fix is to make an explicit copy:

````python
my_list = [1, 2, 3]
b = my_list.copy()   # b is a NEW list with the same values

my_list[1] = 5

print(my_list)   # [1, 5, 3]
print(b)         # [1, 2, 3]  ← b is unaffected
````

> **Rule of thumb:** When you want to preserve the original list while working
> with a modified version, always use `.copy()` (or `list(original)`).

Strings, by contrast, are **immutable** — you cannot change a character in
place:

````python
my_str = 'abc'
# my_str[1] = 'd'   # ← uncomment to see TypeError: 'str' object does not support item assignment
````

The `.append()` method modifies the list **in place** and returns `None`.
A common mistake is storing the return value, expecting to get back the
modified list:

````python
mylist  = [1, 2, 3]
mylist2 = mylist.append(4)    # append returns None

print(mylist2)   # None  ← not what most beginners expect
print(mylist)    # [1, 2, 3, 4]  ← the original was modified
````

**Corrected pattern:** do not assign the result of `.append()`.
Just call it for its side effect:

````python
mylist = [1, 2, 3]
mylist.append(4)    # no assignment
print(mylist)       # [1, 2, 3, 4]
````

## 2 · Tuples

A **tuple** is like a list but **immutable** — once created, you cannot change
its elements. Tuples are written with round brackets `( )`.

Use a tuple when the data should not change:
- GPS coordinates: `(49.2827, -123.1207)` — latitude/longitude of Vancouver
- A record row from a table: `('wheat', 2023, 4.52)`
- Return multiple values from a function (Python packs them as a tuple)

````python
crop_record = ('wheat', 2023, 4.52)   # (crop, year, yield_t_per_ha)
print(crop_record)
print(type(crop_record))              # <class 'tuple'>
````

Indexing and slicing work the same as with lists:

````python
print(crop_record[0])    # 'wheat'
print(crop_record[1:])   # (2023, 4.52)
````

Trying to change a tuple element raises a `TypeError`:

````python
# crop_record[2] = 5.0   # ← uncomment to see TypeError
# Tuples are immutable — use a list if you need to modify the data.
````

### Multiple assignment (tuple unpacking)

You can unpack a tuple (or list) directly into named variables.
This is one of Python's most readable features:

````python
# Each variable on the left receives the matching element from the tuple:
crop, year, yld = crop_record    # crop='wheat', year=2023, yld=4.52
print(crop)   # 'wheat'
print(year)   # 2023
print(yld)    # 4.52
````

````python
# Useful for coordinate pairs:
lat, lon = 49.2827, -123.1207
print(f'Latitude: {lat}, Longitude: {lon}')
````

## 3 · Dictionaries

A **dictionary** maps **keys** to **values**. Think of it like a real
dictionary: you look up a word (the key) to find its definition (the value).

In Python, dictionaries are written with curly braces `{ }`:

```python
d = {key1: value1, key2: value2, ...}
```

Keys must be unique and immutable (strings, numbers, tuples).
Values can be anything — including lists or other dictionaries.

**Example:** storing midterm marks for a set of students.

````python
midterm_marks = {
    'Olivier': 86,
    'Juan':    95,
    'Yasi':    92,
    'Xiao':    97,
    'Janelle': 89,
    'Patrick': 78    # Patrick is now in the dictionary
}
midterm_marks
````

### Accessing values

Use the same square-bracket notation as lists, but with a **key** instead of
an index:

````python
midterm_marks['Yasi']   # look up the value stored under the key 'Yasi'
````

### Navigating all keys and values

`.keys()`, `.values()`, and `.items()` let you inspect the whole dictionary:

````python
midterm_marks.keys()   # returns a view of all keys (student names)
````

````python
midterm_marks.values()   # returns a view of all values (mark numbers)
````

`.items()` returns key-value **pairs** — very useful when you need both in a loop:

````python
midterm_marks.items()   # returns (key, value) pairs — useful for looping over both at once
````

````python
list(midterm_marks.items())   # convert to a regular list of tuples
````

### Adding and updating entries

Assigning to a key that does not exist **adds** a new entry.
Assigning to a key that already exists **updates** its value:

````python
midterm_marks['Tyler'] = 90    # add a new student
midterm_marks
````

### Safe access with `.get()`

If you try to access a key that does not exist, Python raises a `KeyError`.
The `.get(key, default)` method avoids this by returning a default value
instead of crashing:

````python
# Direct access — raises KeyError if the key is missing:
# midterm_marks['Alice']   # ← uncomment to see KeyError

# Safe access with .get():
mark = midterm_marks.get('Alice', 'Student not found')
print(mark)   # 'Student not found' — no crash
````

````python
# .get() also works for present keys:
print(midterm_marks.get('Juan', 'Not found'))   # 95
````

### Sorting a dictionary

`sorted()` on a dictionary iterates over its **keys** and returns them sorted:

````python
sorted(midterm_marks)   # returns sorted list of keys
````

````python
sorted(midterm_marks.values())   # returns sorted list of values
````

### Non-unique keys

Python **silently overwrites** a key if you define it twice.
The second definition wins:

````python
bad = {'A': 1, 'B': 2, 'A': 3}   # 'A' defined twice
bad
````

````python
bad['A']   # 3 — the second definition overwrote the first
````

> **Rule:** If you expect unique keys but get duplicates, the later value
> silently replaces the earlier one. Always validate your data sources.

## 4 · String Methods for Data Cleaning

Real-world data files are messy — commodity prices scraped from a website,
country names from a spreadsheet, or FAOSTAT exports all arrive with
inconsistent capitalisation, stray spaces, or awkward formatting.

Python strings have dozens of built-in methods for cleaning and parsing text.
Here are the ones you will use most often:

### Case conversion and whitespace stripping

**Whitespace** is any invisible character: spaces, tabs, and newlines.
Data exported from spreadsheets often has accidental leading or trailing spaces
that silently break comparisons — `.strip()` removes them.

**Case methods** standardise capitalisation so that `'Wheat'`, `'WHEAT'`, and
`'wheat'` all become the same string:

````python
raw = '  Wheat Futures Price (USD/tonne)  '

print(raw.strip())          # remove leading/trailing whitespace
print(raw.lower())          # all lower-case
print(raw.upper())          # all upper-case
print(raw.strip().lower())  # chain methods: strip first, then lower
````

### Splitting a string into a list

`.split(sep)` divides a string at every occurrence of `sep` and returns a
list. This is how you parse a CSV (comma-separated values) line:

````python
csv_line = 'Canada,wheat,2023,45.6'
parts = csv_line.split(',')
print(parts)          # ['Canada', 'wheat', '2023', '45.6']
print(parts[2])       # '2023'
````

### Joining a list into a string

`.join()` is the reverse of `.split()` — it concatenates a list of strings
with a separator:

````python
fields = ['Canada', 'wheat', '2023', '45.6']
rejoined = ','.join(fields)
print(rejoined)

# Useful for writing rows back to CSV, or building formatted output:
sentence = ' '.join(['Water', 'footprint', 'analysis'])
print(sentence)
````

### Replacing substrings

`.replace(old, new)` returns a **new string** with every occurrence of `old`
replaced by `new`. The original string is not modified (strings are immutable).
You can chain multiple `.replace()` calls to clean several patterns in one step:

````python
messy = 'beef_cattle_GHG_2022.csv'
clean = messy.replace('_', ' ').replace('.csv', '')
print(clean)   # 'beef cattle GHG 2022'
````

### Checking string content

These methods return `True` or `False` and are useful for validating data
before you try to parse or convert it:

| Method | Returns True when… |
|---|---|
| `.isnumeric()` | every character is a digit |
| `.startswith(prefix)` | string begins with `prefix` |
| `.endswith(suffix)` | string ends with `suffix` |
| `.isalpha()` | every character is a letter |
| `.isspace()` | string contains only whitespace |

````python
s = '  3.14  '
print(s.strip().replace('.', '', 1).isnumeric())   # True — can be parsed as float

crop_code = 'WH04'
print(crop_code.startswith('WH'))   # True — a wheat code
print(crop_code.endswith('04'))     # True
````

## 5 · Iteration Patterns

Part A introduced `for` loops over a sequence. Here we add three powerful
patterns that make looping over data more expressive and concise.

### `enumerate` — loop with both index and value

When you need to know the **position** of each element while looping, use
`enumerate`. It wraps any iterable and yields `(index, value)` pairs:

````python
crops = ['wheat', 'corn', 'soybeans', 'rice']

# enumerate(crops) produces: (0,'wheat'), (1,'corn'), (2,'soybeans'), (3,'rice')
# i  receives the index (0, 1, 2, …)
# crop  receives the element ('wheat', 'corn', …)
for i, crop in enumerate(crops):
    print(f'{i}: {crop}')   # f-string: {} is replaced by the variable value
````

`enumerate` is far cleaner than the `range(len(...))` pattern you may have
seen. Compare:

```python
# Old pattern — verbose and error-prone:
for i in range(len(crops)):
    print(i, crops[i])

# Preferred pattern — clear and Pythonic:
for i, crop in enumerate(crops):
    print(i, crop)
```

````python
# MFRE use case: label a table of yield data
yields_t_ha = [3.4, 9.1, 2.8, 4.5]

# zip(crops, yields_t_ha) pairs each crop with its yield:
#   ('wheat', 3.4), ('corn', 9.1), ('soybeans', 2.8), ('rice', 4.5)
# enumerate adds a row index: (0, ('wheat',3.4)), (1, ('corn',9.1)), ...
for i, (crop, yld) in enumerate(zip(crops, yields_t_ha)):
    # :12s means 'pad the string to 12 characters wide' — aligns the columns
    print(f'Row {i}: {crop:12s} {yld} t/ha')
````

### `zip` — loop over two lists in parallel

`zip(list_a, list_b)` pairs up elements by position — like a zipper.
It stops when the shorter list runs out.

````python
countries = ['Canada', 'USA', 'Australia', 'Russia']
wheat_export_Mt = [22.0, 26.5, 21.0, 33.0]   # million tonnes, approx 2023

for country, volume in zip(countries, wheat_export_Mt):
    print(f'{country}: {volume} Mt')
````

````python
# dict() converts a list of (key, value) pairs into a dictionary
# zip(countries, wheat_export_Mt) creates those pairs: ('Canada',22.0), ('USA',26.5), ...
export_dict = dict(zip(countries, wheat_export_Mt))
print(export_dict)
````

### List comprehensions — concise list building

A **list comprehension** is a compact way to build a list from an iterable,
optionally with a filter. The syntax is:

```python
[expression  for item in iterable]
[expression  for item in iterable  if condition]
```

Compare the two approaches:

````python
# Traditional for-loop approach:
yields_kg_ha = []            # start with an empty list
for y in yields_t_ha:
    yields_kg_ha.append(y * 1000)   # convert t/ha → kg/ha and add to list
print(yields_kg_ha)

# List comprehension — one line, same result:
# [expression   for variable in iterable]
#  y * 1000      y             yields_t_ha
yields_kg_ha2 = [y * 1000 for y in yields_t_ha]
print(yields_kg_ha2)
````

````python
# Comprehension with a filter:
# [expression   for variable in iterable   if condition]
# Only tuples where yld > 4 are included in the result list
high_yield = [(crop, yld) for crop, yld in zip(crops, yields_t_ha) if yld > 4]
print(high_yield)
````

````python
# enumerate(repeat) yields (index, value) pairs: (0,14),(1,15),(2,16),(3,16),(4,17)
# The if condition keeps only pairs where the value is 16
repeat = [14, 15, 16, 16, 17]
all_positions = [i for i, val in enumerate(repeat) if val == 16]
print(all_positions)   # [2, 3]
````

### Dictionary comprehensions

The same compact syntax works for dictionaries:

```python
{key_expr: val_expr  for item in iterable  if condition}
```

````python
# Build a dict from two lists so we can use .items() in the comprehension:
yield_dict = dict(zip(crops, yields_t_ha))   # {'wheat':3.4, 'corn':9.1, ...}

# Dict comprehension:  {key_expr: value_expr   for key, val in iterable}
yield_kg = {crop: yld * 1000 for crop, yld in yield_dict.items()}
print(yield_kg)
````

**FizzBuzz** — a classic programming exercise:
Print each number 1–15, but print "Fizz" if divisible by 3,
"Buzz" if divisible by 5, and "FizzBuzz" if divisible by both.

````python
data = list(range(1, 16))   # range(1,16) generates 1,2,3,...,15; list() converts it to a list

for i in data:
    if i % 15 == 0:     # divisible by both 3 and 5 (3×5=15) → check this FIRST
        print('FizzBuzz')
    elif i % 3 == 0:    # divisible by 3 only
        print('Fizz')
    elif i % 5 == 0:    # divisible by 5 only
        print('Buzz')
    else:               # not divisible by either — just print the number
        print(i)
````

An alternative, equivalent way to write FizzBuzz uses **nested** `if` statements
instead of the combined `elif` with `and`. Both approaches are correct — the
nested version makes the logic more explicit:

````python
for i in data:
    if i % 3 == 0:
        if i % 5 == 0:
            print('FizzBuzz')   # divisible by both 3 and 5
        else:
            print('Fizz')       # divisible by 3 only
    elif i % 5 == 0:
        print('Buzz')           # divisible by 5 only
    else:
        print(i)
````

The `and` / `or` operators let you combine conditions:

```python
True and False  →  False
True or  False  →  True
```

````python
print(1 % 2 == 0)              # False  (1 is odd)
print(1 % 2 == 0 or  2 % 2 == 0)   # True   (at least one is even)
print(1 % 2 == 0 and 2 % 2 == 0)   # False  (not both are even)
print(not 1 % 2 == 0)          # True   (NOT flips the boolean)
````

## 7 · Error Handling with `try / except`

When your code reads data from a file or an API, values are often stored as
strings — and not all of them convert cleanly to numbers. Instead of crashing
on bad input, you can **handle the error gracefully** with `try / except`.

````python
# This crashes with ValueError if the string isn't a valid number:
# float('N/A')   # ← uncomment to see the error
````

The `try / except` block:

```python
try:
    # code that might fail
except SomeError:
    # what to do if it fails
```

Execution enters the `try` block; if an exception is raised it jumps
immediately to the `except` block.

````python
def safe_float(value):
    '''
    Attempts to convert value to float.
    Returns None if the conversion fails (e.g. 'N/A', '', 'missing').
    '''
    try:
        return float(value)          # try to convert; if it works, return the float immediately
    except (ValueError, TypeError):  # ValueError: string can't be parsed;  TypeError: wrong type (e.g. None)
        return None                  # conversion failed — return None as a sentinel for "missing"


# Test with a mix of good and bad values:
test_values = ['3.50', 'N/A', '4.20', '', '1.80', 'missing']
# The list comprehension calls safe_float() on every element:
cleaned = [safe_float(v) for v in test_values]
print(cleaned)   # [3.5, None, 4.2, None, 1.8, None]
````

**MFRE application:** FAOSTAT and World Bank datasets often encode missing
values as `'..'`, `'N/A'`, or an empty string. The `safe_float` function above
is a simple guard you can drop into any data-cleaning pipeline.

````python
# An extended version that prints a warning instead of silently returning None:
def safe_float_v2(value, label='value'):
    try:
        return float(value)          # attempt conversion
    except (ValueError, TypeError):
        # !r uses the repr() format — wraps strings in quotes so you can tell it is a string
        print(f'Warning: could not convert {label}={value!r} to float — returning None')
        return None                  # still return None so the calling code can check for it


safe_float_v2('N/A', label='wheat_price_2020')   # label names the variable for the warning message
````

## 8 · Functions Extended

You wrote functions in Part A. Here we look at two important issues that arise
when functions work with **mutable** arguments.

### Function anatomy with a docstring

Every function you write should have a **docstring** — a short description
wrapped in triple quotes immediately after the `def` line. Docstrings make
your code self-documenting and are shown by `help()`.

````python
def my_function(my_argument_a, my_argument_b):
    '''
    One-line summary of what the function does.

    Parameters
    ----------
    my_argument_a : type
        What this argument represents.
    my_argument_b : type
        What this argument represents.

    Returns
    -------
    type
        What is returned.
    '''
    result = my_argument_a + my_argument_b
    return result


help(my_function)
````

### Scope: functions live in their own namespace

A variable defined **inside** a function is local — it does not exist outside:


````python
x = 55   # global variable

def my_func(z):
    x = z * 2    # this 'x' is LOCAL — it does not overwrite the global x
    return x

result = my_func(10)
print(result)   # 20
print(x)        # 55 — the global x is unchanged
````

### The mutable argument trap

Lists (and other mutable objects) are passed to functions **by reference**.
This means the function can modify the original list — even without returning
anything:

````python
def my_func(mylist):
    mylist.append(5)   # modifies the list in place
    return 12


x = [1, 2, 3, 4]
my_func(x)             # returns 12, but also modifies x!
print(x)               # [1, 2, 3, 4, 5]  ← x was changed!
````

To avoid accidentally modifying the caller's list, work on a **copy** inside
the function:

````python
def safe_append(mylist, value):
    '''Returns a new list with value appended; does not modify the original.'''
    local_copy = mylist.copy()   # defensive copy
    local_copy.append(value)
    return local_copy


original = [1, 2, 3, 4]
new_list = safe_append(original, 5)

print('original:', original)   # unchanged
print('new_list:', new_list)   # [1, 2, 3, 4, 5]
````

### Selecting and summarising data with pandas

Once you have a DataFrame, four operations cover the vast majority of
day-to-day work: selecting rows/columns by **position** (`iloc`), selecting
by **label** (`loc`), **filtering** rows on a condition, and
**grouping + aggregating** to compute summaries.

We will use this small crop-export dataset throughout:

````python
import pandas as pd

# This dataset will be used for ALL iloc/loc/filter/groupby examples below.
# 6 rows × 5 columns: two commodities per country
df = pd.DataFrame({
    'country':   ['Canada',  'Canada',  'USA',   'USA',   'Australia', 'Australia'],
    'crop':      ['wheat',   'canola',  'corn',  'soy',   'wheat',     'beef'],
    'year':      [2023,      2023,      2023,    2023,    2023,        2023],
    'export_Mt': [22.0,      5.1,       26.5,    16.3,    21.0,        1.8],
    'price_usd': [280,       550,       215,     480,     275,         4200],
})

df   # display the table in the notebook
````

### `iloc` : select by **position** (integer location)

`iloc[row_index, column_index]` works exactly like indexing a list.
Both rows and columns are addressed by their integer position (0-based).

````python
# iloc[row] — select a single row by its integer position (0 = first row)
# Returns a pandas Series — one value per column
df.iloc[0]
````

````python
# iloc[start:stop] — select a range of rows by position
# Like list slicing: stop is EXCLUDED — iloc[0:3] gives rows 0, 1, 2
df.iloc[0:3]
````

````python
# iloc[row, col] — select a single cell by (row position, column position)
# Row 2 = USA corn row;  column 3 = export_Mt (0=country,1=crop,2=year,3=export_Mt,4=price_usd)
df.iloc[2, 3]
````

````python
# Negative indices count from the end — same as list indexing
# iloc[-2:] = from the second-to-last row to the end
df.iloc[-2:]
````

### `loc` :select by **label** (row index or column name)

`loc[row_label, column_name]` uses the DataFrame's index and column names
rather than integer positions. For most DataFrames the row index is also an
integer, but `loc` treats it as a **label** — so `loc[0:2]` includes row 2
(unlike `iloc`, which excludes the stop).

````python
# loc[:, 'crop'] — ':' means ALL rows;  'crop' is the column name (a label, not a number)
# Returns a pandas Series containing just the 'crop' column
df.loc[:, 'crop']
````

````python
# Pass a LIST of column names to loc to select multiple columns at once
# Returns a DataFrame (still a table), not a Series
df.loc[:, ['country', 'crop', 'export_Mt']]
````

````python
# loc[1:3, [...]] — select rows with labels 1, 2, 3 AND only those two columns
# Note: loc INCLUDES the stop label (row 3 is included) — unlike iloc which excludes it
df.loc[1:3, ['country', 'export_Mt']]
````

### Boolean filtering: selecting rows by condition

The most common way to query a DataFrame is to pass a boolean condition inside
square brackets. This returns only the rows where the condition is `True`.

````python
# df['crop'] == 'wheat' creates a boolean Series: True for wheat rows, False for others
# Passing that Series inside df[...] keeps only the True rows
df[df['crop'] == 'wheat']
````

````python
# Same idea but with a numeric condition:
# df['export_Mt'] > 15 → True for rows where exports exceed 15 Mt
df[df['export_Mt'] > 15]
````

````python
# Combining two conditions with & (AND — both must be True):
# IMPORTANT: wrap each condition in its own parentheses — without them Python mis-parses the expression
df[(df['country'] == 'Canada') & (df['export_Mt'] > 10)]
````

````python
# .query() accepts a condition as a plain string — can be easier to read than bracket syntax
# Use double-quotes around the whole string and single-quotes for string values inside
df.query("crop == 'wheat' and export_Mt > 20")
````

### `groupby` and `agg` — split, apply, combine

`groupby` splits the DataFrame into groups by one or more columns, then
`agg` (short for aggregate) applies a summary function to each group.
This is the pandas equivalent of a pivot table.

````python
# groupby('country') splits the DataFrame into 3 sub-tables (one per country)
# ['export_Mt'] selects that column in each sub-table
# .sum() adds up export_Mt within each group
df.groupby('country')['export_Mt'].sum()
````

````python
# .agg() lets you request several statistics in a single call
# Pass a list of function names as strings: 'sum', 'mean', 'max', 'min', 'count', 'std', etc.
df.groupby('country')['export_Mt'].agg(['sum', 'mean', 'max'])
````

````python
# Grouping by TWO columns creates one row per unique (country, crop) combination
# Returns a Series with a multi-level index (country → crop)
df.groupby(['country', 'crop'])['price_usd'].mean()
````

````python
# Named aggregations — each keyword argument becomes an output column:
#   output_column_name = ('source_column', 'aggregation_function')
# This is the clearest way to build a custom summary table
summary = df.groupby('country').agg(
    total_exports   = ('export_Mt',  'sum'),    # sum of export_Mt per country
    avg_price       = ('price_usd',  'mean'),   # mean of price_usd per country
    num_commodities = ('crop',       'count')   # count of crop rows per country
)
summary
````

> **Mental model for `groupby`:**
>
> 1. **Split** — divide the DataFrame into one sub-table per group value
> 2. **Apply** — run the aggregation function on each sub-table
> 3. **Combine** — stack the results back into a single output table
>
> In August you will use this pattern constantly: group by country or year,
> then sum, average, or count a production or trade variable.
